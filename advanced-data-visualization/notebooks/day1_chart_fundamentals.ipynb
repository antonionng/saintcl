The copilot does not see your matplotlib figure. You will describe your chart in words. Forcing yourself to describe a chart in plain language is part of the lesson - if you cannot describe what the chart shows, the chart probably is not earning its space.

````python
import pathlib
import sys

helper_dir = (pathlib.Path.cwd().resolve().parent.parent / "notebook-helpers")
if str(helper_dir) not in sys.path:
    sys.path.insert(0, str(helper_dir))

import saintagi_copilot as co

top_branch = branches.sort_values("revenue_sar", ascending=False).iloc[0]
bottom_branch = branches.sort_values("revenue_sar", ascending=True).iloc[0]

chart_description = (
    "Chart type: horizontal bar chart, sorted descending by branch revenue.\n"
    "X axis: revenue_sar in SAR (no scientific notation, thousands separators).\n"
    "Y axis: branch name, ordered highest revenue at the top.\n"
    f"Highest revenue branch: {top_branch['branch_name']} at {int(top_branch['revenue_sar']):,} SAR.\n"
    f"Lowest revenue branch: {bottom_branch['branch_name']} at {int(bottom_branch['revenue_sar']):,} SAR.\n"
    f"Total branches plotted: {len(branches)}.\n"
    "Visual hierarchy: top 3 highlighted in a stronger colour, bottom 3 in a muted colour, "
    "the middle group in light grey.\n"
    "Insight title: states the gap between top and bottom revenue, not just the topic.\n"
    "Audience: AJB executive committee, 8 seconds of attention, viewing on a printed deck and on screen."
)

rubric = (
    "You are an AJB executive committee reviewer. Critique this chart "
    "description against five lenses, in this order: 1) is the chart type the "
    "right one for this comparison and audience, 2) is the insight title doing "
    "real work or is it a topic label, 3) is the visual hierarchy directing "
    "the eye to the right thing first, 4) is anything missing that an exec "
    "would ask for in 8 seconds (units, time period, baseline, peer group), "
    "5) one concrete edit that would make the chart more decision-ready. End "
    "with a single sentence: would you put this chart in the steering deck "
    "as is, edit, or replace? Do not invent figures."
)

review = co.critique(
    chart_description,
    rubric,
    exercise_id="m6-d1-chart-critique",
    max_tokens=800,
)

print(review.output if review.output else f"[no output] {review.error_message or ''}")
````
