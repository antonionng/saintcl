```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import numpy as np

plt.rcParams.update({
    'figure.figsize': (10, 6),
    'axes.spines.top': False,
    'axes.spines.right': False,
    'font.size': 11
})

branches = pd.read_csv('../data/branch_performance.csv')
customers = pd.read_csv('../data/customer_metrics.csv')
regions = pd.read_csv('../data/regional_data.csv')
```

```python
# --- Lab 1 Solution: Branch Revenue Comparison ---
sorted_b = branches.sort_values('revenue_sar', ascending=True)
mean_rev = branches['revenue_sar'].mean()

colours = []
for i, val in enumerate(sorted_b['revenue_sar']):
    rank = len(sorted_b) - i
    if rank <= 3:
        colours.append('#2e7d32')
    elif i < 3:
        colours.append('#c62828')
    else:
        colours.append('#9e9e9e')

fig, ax = plt.subplots(figsize=(10, 7))
ax.barh(sorted_b['branch_name'], sorted_b['revenue_sar'] / 1e6, color=colours)
ax.axvline(x=mean_rev / 1e6, color='#1565c0', linestyle='--', linewidth=1, label=f'Mean: SAR {mean_rev/1e6:.1f}M')
ax.set_xlabel('Revenue (SAR Millions)')
ax.set_title('Dhahran Mall and King Fahd Road lead AJB branch revenue;\nTabuk and Al Nakheel trail the network average',
             fontsize=13, fontweight='bold', loc='left')
ax.legend(frameon=False)
plt.tight_layout()
plt.show()
```

```python
# --- Lab 4 Solution: Customer Segment Scatter ---
fig, ax = plt.subplots(figsize=(10, 7))

size_scale = customers['customer_count'] / 80
scatter = ax.scatter(
    customers['churn_rate'], customers['digital_adoption_pct'],
    s=size_scale, c=customers['revenue_per_customer'],
    cmap='YlOrRd', edgecolors='#424242', linewidth=0.5, alpha=0.85
)

for _, row in customers.iterrows():
    ax.annotate(row['segment'], (row['churn_rate'], row['digital_adoption_pct']),
                textcoords='offset points', xytext=(8, 4), fontsize=9)

cbar = plt.colorbar(scatter, ax=ax)
cbar.set_label('Revenue per Customer (SAR)')
ax.set_xlabel('Churn Rate (%)')
ax.set_ylabel('Digital Adoption (%)')
ax.set_title('Youth segment has highest churn (8.5%) despite leading digital adoption;\nCorporate and Premium show low churn with high revenue per customer',
             fontsize=12, fontweight='bold', loc='left')
plt.tight_layout()
plt.show()
```

```python
# --- Lab 5 Solution: Geospatial Bubble Map ---
fig, ax = plt.subplots(figsize=(10, 8))

bubble_sizes = regions['customers'] / 100
scatter = ax.scatter(
    regions['longitude'], regions['latitude'],
    s=bubble_sizes, c=regions['growth_pct'],
    cmap='RdYlGn', edgecolors='#424242', linewidth=0.8, alpha=0.8
)

for _, row in regions.iterrows():
    ax.annotate(f"{row['region']}\n{row['growth_pct']:.1f}% growth",
                (row['longitude'], row['latitude']),
                textcoords='offset points', xytext=(12, 0),
                fontsize=9, fontweight='bold')

cbar = plt.colorbar(scatter, ax=ax)
cbar.set_label('Growth Rate (%)')
ax.set_xlabel('Longitude')
ax.set_ylabel('Latitude')
ax.set_title('Eastern region leads AJB growth at 8.3%;\nNorthern region lags at 2.8% with smallest customer base',
             fontsize=12, fontweight='bold', loc='left')
plt.tight_layout()
plt.show()
```

```python
# --- Lab 7 Solution: Executive Summary KPIs + Primary Chart ---
total_revenue = branches['revenue_sar'].sum()
avg_sat = branches['satisfaction_score'].mean()
top_region = regions.loc[regions['growth_pct'].idxmax()]
top_churn = customers.loc[customers['churn_rate'].idxmax()]

print('=' * 50)
print('EXECUTIVE SUMMARY -- AJB RETAIL BANKING')
print('=' * 50)
print(f"  Total Revenue:     SAR {total_revenue / 1e6:,.1f}M")
print(f"  Avg Satisfaction:  {avg_sat:.1f} / 5.0")
print(f"  Top Growth:        {top_region['region']} ({top_region['growth_pct']:.1f}%)")
print(f"  Highest Churn:     {top_churn['segment']} ({top_churn['churn_rate']:.1f}%)")
print('=' * 50)

target_revenue = 14_000_000
sorted_b = branches.sort_values('revenue_sar', ascending=True)
colours = ['#2e7d32' if v >= target_revenue else '#c62828' for v in sorted_b['revenue_sar']]

fig, ax = plt.subplots(figsize=(10, 7))
ax.barh(sorted_b['branch_name'], sorted_b['revenue_sar'] / 1e6, color=colours)
ax.axvline(x=target_revenue / 1e6, color='#1565c0', linestyle='--', linewidth=1.5,
           label=f'Target: SAR {target_revenue/1e6:.0f}M')
ax.set_xlabel('Revenue (SAR Millions)')
ax.set_title('10 of 15 branches exceed SAR 14M target;\nTabuk, Al Nakheel, and Khamis Mushait need focused intervention',
             fontsize=13, fontweight='bold', loc='left')
ax.legend(frameon=False, loc='lower right')
plt.tight_layout()
plt.show()

print('\nRecommendation: Prioritise digital channel investment in Northern and Southern')
print('branches to close the revenue gap, and launch a Youth retention programme to')
print('address the 8.5% churn rate in AJB\'s fastest-growing digital segment.')
```
