```python
# --- Exercise: Value Case Calculation ---
#
# Build a simple value case for your top opportunity.
# Fill in the variables below with your estimates.

# Assumptions (adjust these for your chosen opportunity)
daily_volume = 800          # units processed per day (complaints, documents, etc.)
time_saved_minutes = 4      # minutes saved per unit with AI
working_days_per_year = 250
fully_loaded_hourly_cost_sar = 120  # SAR per hour, fully loaded

# Calculation
hours_saved_per_year = (daily_volume * time_saved_minutes * working_days_per_year) / 60
annual_saving_sar = hours_saved_per_year * fully_loaded_hourly_cost_sar

print(f"Hours saved per year: {hours_saved_per_year:,.0f}")
print(f"Annual saving (SAR): {annual_saving_sar:,.0f}")

# Conservative estimate (reduce time saved by 25%)
conservative_saving = (daily_volume * (time_saved_minutes * 0.75) * working_days_per_year) / 60 * fully_loaded_hourly_cost_sar
print(f"Conservative annual saving (SAR): {conservative_saving:,.0f}")
print(f"Range: SAR {conservative_saving:,.0f} to SAR {annual_saving_sar:,.0f}")
```
