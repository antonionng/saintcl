````python
import pathlib
import sys

helper_dir = (pathlib.Path.cwd().resolve().parent.parent / "notebook-helpers")
if str(helper_dir) not in sys.path:
    sys.path.insert(0, str(helper_dir))

import saintclaw_copilot as co

shortlist = (
    opportunities.sort_values("priority_score", ascending=False)
    .head(3)[["opportunity_id", "department", "ai_technique", "priority_score"]]
    .to_string(index=False)
)

artifact = (
    "Top 3 AJB AI opportunities I am proposing to fund this quarter:\n"
    f"{shortlist}\n\n"
    "Rationale: ranked by priority_score = value + feasibility - risk. "
    "I believe these are the strongest candidates given current AJB constraints "
    "around regulator review, branch coverage, and analyst capacity."
)

rubric = (
    "You are a senior AJB AI strategy reviewer. Critique this shortlist against "
    "five lenses, in this order: 1) regulator alignment (SAMA, SDAIA, PDPL), "
    "2) data readiness inside a Saudi retail bank, 3) hidden execution risk, "
    "4) measurable customer or efficiency value within 12 months, 5) what is "
    "missing from the rationale that a sponsor would ask. End with one concrete "
    "challenge per opportunity. Do not invent figures."
)

review = co.critique(
    artifact,
    rubric,
    exercise_id="m4-d1-shortlist-critique",
    max_tokens=900,
)

print(review.output if review.output else f"[no output] {review.error_message or ''}")
````
